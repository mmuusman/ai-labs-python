Implementation:
The program below models the environment as a class, implements the agent as a function, and measures
performance: +10 for every square cleaned and −1 for every move.

```python
import random

class VacuumEnvironment:

    def __init__(self):
        self.status = {
            'A': random.randint(0, 1),
            'B': random.randint(0, 1)
        }
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location,
                self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'


def reflex_vacuum_agent(percept):

    location, dirty = percept

    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

random.seed(7)

env = VacuumEnvironment()
env.status = {'A': 1, 'B': 1}
env.agent_location = 'B'

print("Initial state :", env.status,
      "| Agent at", env.agent_location)

score = 0

for step in range(1, 5):
    p = env.percept()
    action = reflex_vacuum_agent(p)

    env.execute(action)

    if action == 'Suck':
        score += 10
    else:
        score -= 1

    print(f"Step {step}: percept={p} -> action={action:5s} | world={env.status}")

print("Final state :", env.status)
print("Performance score:", score)
```

Observation: In Step 4 both squares are already clean, yet the agent still moves Right and loses a point. A
simple reflex agent has no memory: it reacts only to the current percept, so it cannot know that its job is
finished.

```python
import random

class VacuumEnvironment:
    def __init__(self):
        self.status = {
            'A': random.randint(0, 1),
            'B': random.randint(0, 1)
        }
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

def reflex_vacuum_agent(percept):
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

def random_vacuum_agent(percept):
    return random.choice(['Left', 'Right', 'Suck'])

random.seed(42)

initial_states = []
for _ in range(10):
    state = {
        'A': random.randint(0, 1),
        'B': random.randint(0, 1),
        'start': random.choice(['A', 'B'])
    }
    initial_states.append(state)

total_random_score = 0
total_reflex_score = 0

print(f"{'Run':<5} | {'Initial State':<22} | {'Random':<8} | {'Reflex':<8}")
print("-" * 50)

for i, state in enumerate(initial_states):
    env_rand = VacuumEnvironment()
    env_rand.status = {'A': state['A'], 'B': state['B']}
    env_rand.agent_location = state['start']

    rand_score = 0
    for _ in range(20):
        p = env_rand.percept()
        action = random_vacuum_agent(p)
        env_rand.execute(action)
        if action == 'Suck':
            rand_score += 10
        else:
            rand_score -= 1
    total_random_score += rand_score

    env_refl = VacuumEnvironment()
    env_refl.status = {'A': state['A'], 'B': state['B']}
    env_refl.agent_location = state['start']

    refl_score = 0
    for _ in range(20):
        p = env_refl.percept()
        action = reflex_vacuum_agent(p)
        env_refl.execute(action)
        if action == 'Suck':
            refl_score += 10
        else:
            refl_score -= 1
    total_reflex_score += refl_score

    state_str = f"A:{state['A']}, B:{state['B']}, Pos:{state['start']}"
    print(f"{i+1:<5} | {state_str:<22} | {rand_score:<8} | {refl_score:<8}")

print("-" * 50)
print(f"{'Average Score':<30} | {total_random_score/10:<8.1f} | {total_reflex_score/10:<8.1f}")
```

4. Model-based agent: Extend the reflex agent so that it remembers which locations it has already
cleaned and performs NoOp (stops) when both locations are clean. Show that its performance
score is higher than the simple reflex agent when moves are penalised.


```python
import random

class VacuumEnvironment:
    def __init__(self):
        self.status = {
            'A': random.randint(0, 1),
            'B': random.randint(0, 1)
        }
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

class ModelBasedVacuumAgent:
    def __init__(self):
        self.model = {'A': None, 'B': None}

    def program(self, percept):
        location, dirty = percept
        self.model[location] = 'Dirty' if dirty else 'Clean'

        if self.model['A'] == 'Clean' and self.model['B'] == 'Clean':
            return 'NoOp'

        if dirty:
            self.model[location] = 'Clean'
            return 'Suck'
        elif location == 'A':
            return 'Right'
        else:
            return 'Left'

random.seed(42)

env = VacuumEnvironment()
env.status = {'A': 1, 'B': 1}
env.agent_location = 'B'

print("Initial state :", env.status, "| Agent at", env.agent_location)

agent = ModelBasedVacuumAgent()
score = 0

for step in range(1, 6):
    p = env.percept()
    action = agent.program(p)

    env.execute(action)

    if action == 'Suck':
        score += 10
    elif action == 'NoOp':
        score += 0
    else:
        score -= 1

    print(f"Step {step}: percept={p} -> action={action:5s} | world={env.status}")

print("Final state :", env.status)
print("Performance score:", score)
```

5. Extend the vacuum world to a 3 × 3 grid of rooms with random dirt. Implement a model-based reflex agent
that visits every room and cleans it. Report the number of moves, the number of cleaned rooms and the
performance score.

This code scales the VacuumEnvironment up to a 3x3 grid (9 rooms labeled from (0,0) to (2,2)). The ModelBasedVacuumAgent3x3 tracks the cleanliness of all rooms in its internal memory (self.model) and keeps track of its own spatial movements to systematically navigate and clean every single grid room.
Once every room is marked clean, it stops using a NoOp command to maximize its performance score.

```python
import random

class VacuumEnvironment3x3:
    def __init__(self):
        self.status = {}
        for r in range(3):
            for c in range(3):
                self.status[(r, c)] = random.randint(0, 1)
        self.agent_location = (random.randint(0, 2), random.randint(0, 2))
        self.moves_count = 0
        self.cleaned_count = 0

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        r, c = self.agent_location
        if action == 'Suck':
            if self.status[(r, c)] == 1:
                self.status[(r, c)] = 0
                self.cleaned_count += 1
        elif action == 'Up' and r > 0:
            self.agent_location = (r - 1, c)
            self.moves_count += 1
        elif action == 'Down' and r < 2:
            self.agent_location = (r + 1, c)
            self.moves_count += 1
        elif action == 'Left' and c > 0:
            self.agent_location = (r, c - 1)
            self.moves_count += 1
        elif action == 'Right' and c < 2:
            self.agent_location = (r, c + 1)
            self.moves_count += 1

class ModelBasedVacuumAgent3x3:
    def __init__(self):
        self.model = {(r, c): None for r in range(3) for c in range(3)}
        self.path = [
            (0, 0), (0, 1), (0, 2),
            (1, 2), (1, 1), (1, 0),
            (2, 0), (2, 1), (2, 2)
        ]

    def program(self, percept):
        location, dirty = percept
        self.model[location] = 'Clean' if not dirty else 'Dirty'

        if all(status == 'Clean' for status in self.model.values()):
            return 'NoOp'

        if dirty:
            self.model[location] = 'Clean'
            return 'Suck'

        current_index = self.path.index(location)
        for next_loc in self.path[current_index + 1:]:
            if self.model[next_loc] is None:
                target_loc = next_loc
                break
        else:
            for loc in self.path:
                if self.model[loc] is None:
                    target_loc = loc
                    break
            else:
                return 'NoOp'

        curr_r, curr_c = location
        tar_r, tar_c = target_loc

        if curr_r < tar_r: return 'Down'
        if curr_r > tar_r: return 'Up'
        if curr_c < tar_c: return 'Right'
        if curr_c > tar_c: return 'Left'
        return 'NoOp'

random.seed(42)

env = VacuumEnvironment3x3()
agent = ModelBasedVacuumAgent3x3()

initial_dirty_count = sum(env.status.values())
print("Initial Environment Grid State (1 = Dirty, 0 = Clean):")
for r in range(3):
    row_str = " | ".join(f"{(r, c)}: {env.status[(r, c)]}" for c in range(3))
    print(f"  {row_str}")
print(f"Agent starting location: {env.agent_location}\n")

score = 0
max_steps = 50

for step in range(1, max_steps + 1):
    p = env.percept()
    action = agent.program(p)

    env.execute(action)

    if action == 'Suck':
        score += 10
    elif action == 'NoOp':
        score += 0
    else:
        score -= 1

    print(f"Step {step:02d}: Location={p[0]} Status={'Dirty' if p[1] else 'Clean'} -> Action={action:5s}")

    if action == 'NoOp':
        print(f"\n[Finished] Agent issued NoOp at Step {step} because all rooms are clean.")
        break

print("\n" + "="*40)
print("FINAL PERFORMANCE REPORT")
print("="*40)
print(f"Total Moves Performed  : {env.moves_count}")
print(f"Total Rooms Cleaned    : {env.cleaned_count} (out of {initial_dirty_count} initially dirty)")
print(f"Final Performance Score: {score}")
print("="*40)
```
